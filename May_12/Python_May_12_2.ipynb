```python
import random

class SlotMachine:
    def __init__(self, initial_coins):

        # 絵柄リスト
        self.symbols = ['🍎', '🍋', '🍒', '💎']

        # 所持コイン
        self.coins = initial_coins

        # 現在のリール:3x3のリールを保存
        self.current_reels = []

    def spin(self, bet):

        # コイン不足チェック
        if self.coins < bet:
            print("コインが足りません！")
            return False

        # ベット分減らす
        self.coins -= bet

        # =========================
        # 3 X 3のリールを作成
        # 各マス目ごとにランダムに
        # 絵柄を抽選する。
        # =========================
        self.current_reels = []
        for row in range(3):   # 行作成を3回繰り返す
          reel_row = []
          for col in range(3):   # 列作成を3回繰り返す
            symbol = random.choice(self.symbols)
            reel_row.append(symbol)
          # 1行完成したら追加
          self.current_reels.append(reel_row)

        return True

    #判定ライン
    def check_result(self):
        #横・縦・斜めを全部ここへ入れる。
        lines = []

        # =========================
        # 横:1行ずつ取り出す
        # =========================
        for row in self.current_reels:
            lines.append(row)

        # =========================
        # 縦:[行][列]を使って楯列を判断
        # =========================
        for col in range(3):

            line = [
                self.current_reels[0][col], #1行目のリール
                self.current_reels[1][col], #2行目のリール
                self.current_reels[2][col]  #3行目のリール
            ]

            lines.append(line)

        # =========================
        # 斜め:[行][列]を使って楯列を判断
        # =========================

        # 左上 → 右下方向
        lines.append([
            self.current_reels[0][0],  #1行目リールの1番目の絵柄
            self.current_reels[1][1],  #2行目リールの2番目の絵柄
            self.current_reels[2][2]   #3行目リールの3番目の絵柄
        ])

        # 右上 → 左下方向
        lines.append([
            self.current_reels[0][2],  #1行目リールの3番目の絵柄
            self.current_reels[1][1],  #2行目リールの2番目の絵柄
            self.current_reels[2][0]   #3行目リールの1番目の絵柄
        ])

        # =========================
        # 当たり判定
        # =========================
        hit_count = 0
        total_payout = 0
        #全ラインを1つずつ確認
        for line in lines:

            # 3つ一致
            if line[0] == line[1] == line[2]:

                hit_count += 1

                # 💎だけ高配当
                if line[0] == '💎':
                    payout = 40
                else:
                    payout = 10
                #絵柄がそろった場合に配当を手持ちコインに追加
                total_payout += payout

        # =========================
        # 配当
        # =========================
        if hit_count > 0:

            self.coins += total_payout

            return f"{hit_count}ライン当たり！ {total_payout}枚獲得！"

        return "ハズレ..."

    def show_status(self):

        print("\n====== SLOT ======")

        for row in self.current_reels:
            print(" | ".join(row))

        print("==================")
        print(f"所持コイン: {self.coins}\n")


# =========================
# ゲーム開始
# =========================
machine = SlotMachine(initial_coins=100)

while machine.coins > 0:

    cmd = input("Enterでスピン (qで終了): ")

    if cmd == 'q':
        break

    if machine.spin(bet=1):

        machine.show_status()

        print(machine.check_result())

print("ゲーム終了")
```
